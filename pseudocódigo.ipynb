```python
INICIO

  expedientes ← lista vacía

  contador ← 0

  REPETIR

    MOSTRAR menú: 1 Registrar, 2 Buscar, 3 Listar por fecha, 4 Eliminar, 5 Salir

    LEER opción

    SEGÚN opción

      1: registrar()

      2: buscar()

      3: listar()

      4: eliminar()

      5: MOSTRAR "Sistema finalizado"

      otro: MOSTRAR "Opción inválida"

  HASTA opción = 5

FIN



FUNCIÓN registrar()

  LEER nombre, apellidos (pasar a mayúsculas), dni, tipo, asunto, fecha, área

  SI algún campo está vacío ENTONCES MOSTRAR error y SALIR de la función

  SI dni no es numérico O longitud(dni) ≠ 8 ENTONCES MOSTRAR error y SALIR

  PARA CADA e EN expedientes

    SI e.dni = dni ENTONCES MOSTRAR "DNI ya registrado" y SALIR

  SI fecha no es una fecha válida AAAA-MM-DD ENTONCES MOSTRAR error y SALIR

  contador ← contador + 1

  codigo ← "EXP-" + contador con 4 dígitos

  AGREGAR a expedientes el registro (codigo, datos, estado "Registrado")

  MOSTRAR el expediente registrado

FIN FUNCIÓN



FUNCIÓN buscar()

  SI expedientes está vacía ENTONCES MOSTRAR aviso y SALIR

  LEER clave (pasar a mayúsculas)

  PARA CADA e EN expedientes

    SI clave = e.codigo O clave = e.dni ENTONCES MOSTRAR e y SALIR

  MOSTRAR "Expediente no encontrado"

FIN FUNCIÓN



FUNCIÓN listar()

  SI expedientes está vacía ENTONCES MOSTRAR aviso y SALIR

  ordenados ← expedientes ordenados por fecha (ascendente)

  PARA CADA e EN ordenados: MOSTRAR e

FIN FUNCIÓN



FUNCIÓN eliminar()

  SI expedientes está vacía ENTONCES MOSTRAR aviso y SALIR

  MOSTRAR código, DNI y nombre de cada expediente

  LEER clave (pasar a mayúsculas)

  PARA CADA e EN expedientes

    SI clave = e.codigo O clave = e.dni ENTONCES

      LEER confirmar

      SI confirmar = "S" ENTONCES QUITAR e de expedientes

      SINO MOSTRAR "Eliminación cancelada"

      SALIR

  MOSTRAR "Expediente no encontrado"

FIN FUNCIÓN
```
